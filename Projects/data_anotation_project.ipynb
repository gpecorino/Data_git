```python
#Function to build grid
def build_grid(x_axis,y_axis,symbols):
    max_x = int(max(x_axis))
    max_y = int(max(y_axis))

    grid = [[' ' for _ in range(max_x + 1)] for _ in range(max_y + 1)]

    for x, y, sym in zip(x_axis,y_axis,symbols):
        grid[y][x] = sym

    for row in reversed(grid):
        print(''.join(row))

def decode(url):
    #call url
    url = url
    response = requests.get(export_url)
    response.raise_for_status()

    #load in data we want from webpage
    soup = BeautifulSoup(response.text, 'html.parser')
    content_div = soup.find('div', class_='doc-content')
    text = content_div.get_text(separator='\n').strip()
    start = text.find('y-coordinate')
    data = text[start+13:].splitlines()

    #make lists of values
    for i in range(len(data)):
        x_axis = [int(data[i]) for i in range(0,len(data),3)]
        symbol = [data[i] for i in range(1,len(data),3)]
        y_axis = [int(data[i]) for i in range(2,len(data),3)]

    build_grid(x_axis,y_axis,symbol)
```
